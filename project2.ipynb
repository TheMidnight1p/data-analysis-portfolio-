```python
# Data Manipulation
```

```python
# Total Sales by Category
Category_Sales = (
    df.groupby('Category')['Sales']
    .sum()
    .sort_values(ascending=False)
)
Category_Sales
```

```python
# Top 10 products by Sale
top_products = (
    df.groupby('Product Name')['Sales']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)
top_products  
```

```python
# Sale by Location
location_sale = (
    df.groupby('Location')['Sales']
    .sum()
    .sort_values(ascending=False)
)
location_sale
```

```python
# Most common payment method
payment_counts = (
    df.groupby('Payment Method').value_counts()
)
payment_counts    
```

```python
# Total quantity sold by Category
quantity_by_category = (
    df.groupby('Quantity')['Category']
    .sum()
    .sort_values(ascending =False)
)
quantity_by_category
```
